## Duplicate Analysis

The training dataset contains 480 duplicate records.

Duplicates may introduce bias during model training because repeated samples can cause the model to memorize specific patterns instead of learning general sentiment representations.

The duplicates will be removed during the data preprocessing stage while preserving the original raw dataset.

```python
# Investigating the duplicates

duplicate_rows = train_df[train_df.duplicated()]

duplicate_rows.head()
```

```python
duplicate_rows['label'].value_counts()
```

## Duplicate Records Analysis

Duplicate sentiment distribution:

| Sentiment | Duplicate Count |

Since duplicate records can introduce bias during model training by over-representing certain samples, they will be removed during the preprocessing phase. The original dataset will remain unchanged in the `dataset/raw` directory to preserve data integrity.

```python
# Verifying Whether the Dataset Is Balanced

train_df['label'].value_counts()
```

## Class Distribution Analysis

The training dataset contains three sentiment classes:

| Sentiment | Count | Percentage |
|-----------|------:|-----------:|
| Negative | 3241 | 63.29% |
| Positive | 1808 | 35.30% |
| Neutral | 72 | 1.41% |

### Observation

The dataset is imbalanced. The Negative class dominates the dataset, while the Neutral class is significantly underrepresented.

### Potential Impact

An imbalanced dataset may bias machine learning models toward predicting the majority class. Therefore, evaluation metrics such as Precision, Recall, and F1-score will be used alongside Accuracy. Class balancing techniques will be considered later if model performance on the Neutral class is unsatisfactory.
