```python
# Imported Library & Load the Processed Dataset

import pandas as pd

processed_df = pd.read_csv("../dataset/processed/train_processed.csv")

processed_df.head()
```

```python
# Check Dataset Information

processed_df.info()
```

```python
processed_df.sample(5)
```

```python
# Check Label Distribution Again

processed_df["label"].value_counts()
```

```python
# Encode the Labels

from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

processed_df["label_encoded"] = label_encoder.fit_transform(
    processed_df["label"]
)
```

```python
processed_df.head()
```

```python
# Verify Encoding

label_encoder.classes_
```

```python
for index, label in enumerate(label_encoder.classes_):
    print(f"{label} --> {index}")
```

```python
# Features 

X = processed_df["final_text"]

# Target

y = processed_df["label_encoded"]

```

```python
print("Feature Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Split the Dataset

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
# Verify the Split

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Labels  :", y_train.shape)
print("Testing Labels   :", y_test.shape)
```

```python
# Confirm Class Distribution

print("Training Set:", )
print(y_train.value_counts())

print("\nTesting Set:")
print(y_test.value_counts())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Initialize the Vectorizer
tfidf_vectorizer = TfidfVectorizer(
    max_features=5000
)
```

```python
# Fit Only on Training Data

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
```

```python
# Transform Test Data

X_test_tfidf = tfidf_vectorizer.transform(X_test)
```

```python
# Check Matrix Shape

print("Training TF-IDF Shape:", X_train_tfidf.shape)
print("Testing TF-IDF Shape:", X_test_tfidf.shape)
```

```python
# Inspect Learned Features

feature_names = tfidf_vectorizer.get_feature_names_out()

feature_names[:50]
```

```python
# Check Total Vocabulary Size

print("Total Features:", len(feature_names))
```
